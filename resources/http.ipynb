```python
import asyncio

import requests
from httpx import AsyncClient

api_url = 'https://pokeapi.co/api/v2/pokemon/'
numbers = range(1, 100)

def get_pokemons():
    for n in numbers:
        response = requests.get(f'{api_url}{n}')
        print(response.json())

async def get_pokemons_async():
    async with AsyncClient() as client:
        responses = await asyncio.gather(*[client.get(f'{api_url}{n}') for n in numbers])
        for response in responses:
            print(response.json())

# get_pokemons()
await get_pokemons_async()
```
